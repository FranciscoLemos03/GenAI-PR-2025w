```python
# download_example_pdfs.py
import os
import requests
from datasets import load_dataset
import arxiv

EXAMPLES_DIR = "example_pdfs_to_upload"
os.makedirs(EXAMPLES_DIR, exist_ok=True)

def get_arxiv_id(result):
    return result.entry_id.split("/")[-1]

def download_pdf(pdf_url, out_path):
    r = requests.get(pdf_url)
    r.raise_for_status()
    with open(out_path, "wb") as f:
        f.write(r.content)

ds = load_dataset("CShorten/ML-ArXiv-Papers", split="train")
gpt_papers = [p for p in ds if "gpt" in p["title"].lower()]

for p in gpt_papers:
    title = p["title"]

    search = arxiv.Search(query=f'ti:"{title}"', max_results=1)
    results = list(arxiv.Client().results(search))
    if not results:
        print("No result found:", title)
        continue

    result = results[0]
    arxiv_id = get_arxiv_id(result) 
    pdf_path = os.path.join(EXAMPLES_DIR, f"{arxiv_id}.pdf")

    if os.path.exists(pdf_path):
        print("Already exists:", arxiv_id)
        continue

    print(f"Downloading: {arxiv_id}")
    download_pdf(result.pdf_url, pdf_path)

print("All example PDFs downloaded!")
```
